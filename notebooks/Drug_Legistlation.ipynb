````python
#Code to make first chart.
File_path_income = "C:\Users\Igaba\OneDrive - Dublin City University\ProjectTemplate\data\Eurostat_EUDA_datasets\Eurostat_EUDA_datasets\eurostat-ilc_di03-mean-median-income.csv"
File_path_annextab6 = "C:\\Users\\Igaba\\OneDrive - Dublin City University\\ProjectTemplate\\data\\Eurostat_EUDA_datasets\\Eurostat_EUDA_datasets\\edr2026-annex-table-6-other-indicators.csv"
a6 = pd.read_csv(File_path_annextab6)
incom = pd.read_csv(File_path_income)
#a6.head()
plt.figure(figsize=(8, 6))
row = ['Ireland', 'Portugal', 'Netherlands']
column = 'Drug-induced deaths - Cases per million population - Aged 15-64'
plot_data = a6[a6['Country'].isin(row)][['Country',column]]
name_to_code = {"Ireland": "IE", "Portugal": "PT", "Netherlands": "NL"}

inc_f = inc[
    (inc["geo"].isin(name_to_code.values())) &
    (inc["unit"] == "EUR") &
    (inc["age"] == "TOTAL") &
    (inc["sex"] == "T") &
    (inc["statinfo"] == "MEAN_EI") &
    (inc["TIME_PERIOD"] == 2022)
].copy()
inc_f["OBS_VALUE"] = pd.to_numeric(inc_f["OBS_VALUE"], errors="coerce")
inc_f["Country"] = inc_f["geo"].map({v: k for k, v in name_to_code.items()})

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

sns.barplot(data=plot_data, x='Country', y=column, palette='pink', ax=ax1)
ax1.set_title('Drug-induced deaths per million')
ax1.set_xlabel('Country')
ax1.set_ylabel('Deaths per million')

sns.barplot(data=inc_f, x='Country', y='OBS_VALUE', palette='lightblue', ax=ax2)
ax2.set_title('Mean Equivalised Income, 2022')
ax2.set_xlabel('Country')
ax2.set_ylabel('Mean Equivalised Income (EUR)')

plt.tight_layout()
plt.show()

#sns.barplot(data=plot_data, x='Country', y=column, palette='pink')
#plt.title('Drug-induced deaths per million')
#plt.xlabel('Country')
#plt.ylabel('Deaths per million')
#plt.show()

````
